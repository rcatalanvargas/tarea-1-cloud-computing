```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
distribucion_split = pd.DataFrame({
    "Entrenamiento (%)": (
        y_train.value_counts(normalize=True).mul(100).round(2)
    ),
    "Prueba (%)": (
        y_test.value_counts(normalize=True).mul(100).round(2)
    ),
})

distribucion_split
```

## Separación entre entrenamiento y prueba

Se utilizó una división estratificada de los datos:

- 80 % para entrenamiento: 40.000 estudiantes.
- 20 % para prueba: 10.000 estudiantes.
- Semilla fija: `random_state=42`.
- Estratificación mediante `stratify=y`.

La proporción de clases se conservó en ambos conjuntos:

- `Placed`: 78,08 %.
- `Not Placed`: 21,92 %.

El conjunto de prueba se mantendrá aislado durante el ajuste del preprocesamiento y la selección del modelo. Se utilizará únicamente para la evaluación final.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocesador = ColumnTransformer(
    transformers=[
        (
            "numericas",
            StandardScaler(),
            columnas_numericas,
        ),
        (
            "categoricas",
            OneHotEncoder(handle_unknown="ignore"),
            columnas_categoricas,
        ),
    ],
    remainder="drop",
)

preprocesador
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

clasificador = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42,
)

pipeline_modelo = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador),
        ("clasificador", clasificador),
    ]
)

pipeline_modelo
```

```python
from sklearn.model_selection import StratifiedKFold

validacion_cruzada = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42,
)

validacion_cruzada
```

```python
metricas_cv = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "f1_macro": "f1_macro",
    "roc_auc": "roc_auc",
}

metricas_cv
```

```python
from sklearn.model_selection import cross_validate

resultados_cv = cross_validate(
    estimator=pipeline_modelo,
    X=X_train,
    y=y_train,
    cv=validacion_cruzada,
    scoring=metricas_cv,
    return_train_score=False,
)

resultados_cv
```

```python
resumen_cv = pd.DataFrame({
    metrica: {
        "media": resultados_cv[f"test_{metrica}"].mean(),
        "desviacion_estandar": resultados_cv[f"test_{metrica}"].std(),
    }
    for metrica in metricas_cv
}).T.round(4)

resumen_cv
```

```python
from sklearn.base import clone

pipeline_sin_balanceo = clone(pipeline_modelo)

pipeline_sin_balanceo.set_params(
    clasificador__class_weight=None
)

resultados_cv_sin_balanceo = cross_validate(
    estimator=pipeline_sin_balanceo,
    X=X_train,
    y=y_train,
    cv=validacion_cruzada,
    scoring=metricas_cv,
    return_train_score=False,
)

resumen_cv_sin_balanceo = pd.DataFrame({
    metrica: {
        "media": resultados_cv_sin_balanceo[f"test_{metrica}"].mean(),
        "desviacion_estandar": resultados_cv_sin_balanceo[f"test_{metrica}"].std(),
    }
    for metrica in metricas_cv
}).T.round(4)

resumen_cv_sin_balanceo
```

## Selección del balanceo de clases

Se comparó la misma regresión logística con y sin `class_weight="balanced"` mediante validación cruzada estratificada de cinco particiones.

| Métrica | Con balanceo | Sin balanceo |
|---|---:|---:|
| Accuracy | 0,7485 | 0,8195 |
| Balanced accuracy | 0,7381 | 0,6518 |
| F1 macro | 0,6905 | 0,6767 |
| ROC-AUC | 0,8114 | 0,8114 |

La configuración sin balanceo obtiene mayor accuracy porque favorece la clase mayoritaria `Placed`. Sin embargo, reduce de forma importante la balanced accuracy y también disminuye el F1 macro.

Se selecciona `class_weight="balanced"` porque el objetivo es evaluar adecuadamente ambas clases, incluida `Not Placed`, que representa el 21,92 % de los casos. La accuracy se reportará como métrica complementaria, pero no se utilizará de manera aislada.

```python
columnas_numericas_sin_employability = [
    columna
    for columna in columnas_numericas
    if columna != "Employability_Score"
]

preprocesador_sin_employability = ColumnTransformer(
    transformers=[
        (
            "numericas",
            StandardScaler(),
            columnas_numericas_sin_employability,
        ),
        (
            "categoricas",
            OneHotEncoder(handle_unknown="ignore"),
            columnas_categoricas,
        ),
    ],
    remainder="drop",
)

pipeline_sin_employability = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador_sin_employability),
        (
            "clasificador",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42,
            ),
        ),
    ]
)

X_train_sin_employability = X_train.drop(
    columns=["Employability_Score"]
)

resultados_cv_sin_employability = cross_validate(
    estimator=pipeline_sin_employability,
    X=X_train_sin_employability,
    y=y_train,
    cv=validacion_cruzada,
    scoring=metricas_cv,
    return_train_score=False,
)

comparacion_employability = pd.DataFrame({
    "Con Employability_Score": resumen_cv["media"],
    "Sin Employability_Score": [
        resultados_cv_sin_employability[f"test_{metrica}"].mean()
        for metrica in metricas_cv
    ],
}).round(4)

comparacion_employability
```

### Decisión sobre `Employability_Score`

Se comparó el mismo pipeline de regresión logística balanceada mediante
validación cruzada estratificada de 5 folds, usando únicamente el conjunto
de entrenamiento.

Los resultados con y sin `Employability_Score` fueron prácticamente
idénticos. Las diferencias en `balanced_accuracy` y `f1_macro` fueron de
solo 0.0001, mientras que `accuracy` y `roc_auc` no cambiaron.

Por lo tanto, se excluye `Employability_Score` del modelo final porque:

- no aporta una mejora material en las métricas;
- es una puntuación compuesta cuya fórmula no está documentada;
- contiene información redundante respecto de otras variables;
- su exclusión simplifica el contrato de entrada de la futura API.

La columna se conserva intacta en el dataset original, pero no será
utilizada como variable predictora.

```python
# Variables definitivas del modelo
variables_predictoras_finales = [
    columna
    for columna in variables_predictoras
    if columna != "Employability_Score"
]

columnas_numericas_finales = [
    columna
    for columna in columnas_numericas
    if columna != "Employability_Score"
]

columnas_categoricas_finales = columnas_categoricas.copy()

# Conservamos exactamente las mismas filas del split original
X_train_final = X_train[variables_predictoras_finales].copy()
X_test_final = X_test[variables_predictoras_finales].copy()

# Preprocesamiento definitivo
preprocesador_final = ColumnTransformer(
    transformers=[
        (
            "numericas",
            StandardScaler(),
            columnas_numericas_finales,
        ),
        (
            "categoricas",
            OneHotEncoder(handle_unknown="ignore"),
            columnas_categoricas_finales,
        ),
    ],
    remainder="drop",
)

# Pipeline definitivo todavía sin entrenar
modelo_final = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador_final),
        (
            "clasificador",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42,
            ),
        ),
    ]
)

print("Variables predictoras finales:", len(variables_predictoras_finales))
print("Numéricas:", len(columnas_numericas_finales))
print("Categóricas:", len(columnas_categoricas_finales))
print("X_train_final:", X_train_final.shape)
print("X_test_final:", X_test_final.shape)
print("\nVariables ordenadas:")
print(variables_predictoras_finales)
```

```python
modelo_final.fit(X_train_final, y_train)

print("Pipeline definitivo entrenado correctamente.")
print(
    "Clases aprendidas:",
    modelo_final.named_steps["clasificador"].classes_,
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    roc_auc_score,
)

# Predicción de clases
y_pred_test = modelo_final.predict(X_test_final)

# Probabilidad correspondiente explícitamente a la clase "Placed"
clases_modelo = list(
    modelo_final.named_steps["clasificador"].classes_
)
indice_placed = clases_modelo.index("Placed")

y_prob_placed_test = modelo_final.predict_proba(
    X_test_final
)[:, indice_placed]

# Target binario únicamente para calcular ROC-AUC
y_test_binario = (y_test == "Placed").astype(int)

metricas_test = {
    "accuracy": accuracy_score(y_test, y_pred_test),
    "balanced_accuracy": balanced_accuracy_score(
        y_test,
        y_pred_test,
    ),
    "f1_macro": f1_score(
        y_test,
        y_pred_test,
        average="macro",
    ),
    "roc_auc": roc_auc_score(
        y_test_binario,
        y_prob_placed_test,
    ),
}

pd.Series(
    metricas_test,
    name="valor_test",
).round(4)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
)

orden_clases = ["Not Placed", "Placed"]

matriz_confusion = pd.DataFrame(
    confusion_matrix(
        y_test,
        y_pred_test,
        labels=orden_clases,
    ),
    index=[
        "Real: Not Placed",
        "Real: Placed",
    ],
    columns=[
        "Predicho: Not Placed",
        "Predicho: Placed",
    ],
)

reporte_clasificacion = pd.DataFrame(
    classification_report(
        y_test,
        y_pred_test,
        labels=orden_clases,
        output_dict=True,
        zero_division=0,
    )
).T.round(4)

display(matriz_confusion)
display(reporte_clasificacion)
```

### Evaluación final sobre el conjunto de prueba

El pipeline definitivo se evaluó una sola vez sobre 10.000 observaciones
reservadas mediante la separación estratificada inicial.

Resultados:

- Accuracy: 0.7414
- Balanced accuracy: 0.7296
- F1 macro: 0.6825
- ROC-AUC: 0.8031

La accuracy no se utiliza como única métrica porque el 78,08% de las
observaciones pertenece a la clase `Placed`. Un clasificador que predijera
siempre esa clase obtendría una accuracy alta, pero no identificaría ningún
caso `Not Placed`.

El modelo identificó correctamente 1.553 de los 2.192 casos `Not Placed`,
alcanzando un recall de 0.7085 para la clase minoritaria. Para la clase
`Placed`, obtuvo un recall de 0.7506 y una precisión de 0.9017.

El balanceo de clases reduce la accuracy global, pero permite distribuir
mejor la capacidad predictiva entre ambas clases. Las métricas del conjunto
de prueba son cercanas a las obtenidas mediante validación cruzada, lo que
indica una generalización razonablemente estable.

```python
import sys
import numpy as np
import pandas as pd
import scipy
import sklearn
import joblib

versiones_entorno = {
    "python": sys.version.split()[0],
    "numpy": np.__version__,
    "pandas": pd.__version__,
    "scipy": scipy.__version__,
    "scikit-learn": sklearn.__version__,
    "joblib": joblib.__version__,
}

pd.Series(
    versiones_entorno,
    name="version",
)
```

```python
from pathlib import Path
import json

# Desde notebooks/ retrocedemos a la raíz y entramos en model/
carpeta_modelo = Path("../model")
ruta_modelo = carpeta_modelo / "model.pkl"
ruta_metadata = carpeta_modelo / "metadata.json"

# Guardar el pipeline completo:
# preprocesamiento + regresión logística entrenada
joblib.dump(
    modelo_final,
    ruta_modelo,
)

# Resumen de la validación cruzada del modelo definitivo
metricas_cv_final = {
    metrica: {
        "media": round(
            float(
                resultados_cv_sin_employability[
                    f"test_{metrica}"
                ].mean()
            ),
            4,
        ),
        "desviacion_estandar": round(
            float(
                resultados_cv_sin_employability[
                    f"test_{metrica}"
                ].std()
            ),
            4,
        ),
    }
    for metrica in metricas_cv
}

metadata = {
    "problem_type": "binary_classification",
    "target": "Placement_Status",
    "positive_class": "Placed",
    "classes": [
        str(clase)
        for clase in modelo_final
        .named_steps["clasificador"]
        .classes_
    ],
    "input_features": variables_predictoras_finales,
    "numeric_features": columnas_numericas_finales,
    "categorical_features": columnas_categoricas_finales,
    "excluded_features": [
        "Student_ID",
        "Academic_Performance",
        "Company_Tier",
        "Career_Field",
        "Placement_Mode",
        "Starting_Salary_USD",
        "Employability_Score",
    ],
    "dataset_rows": len(X),
    "training_rows": len(X_train_final),
    "test_rows": len(X_test_final),
    "data_split": {
        "test_size": 0.20,
        "random_state": 42,
        "stratified": True,
    },
    "cross_validation": {
        "method": "StratifiedKFold",
        "n_splits": 5,
        "shuffle": True,
        "random_state": 42,
    },
    "estimator": {
        "name": "LogisticRegression",
        "max_iter": 1000,
        "class_weight": "balanced",
        "random_state": 42,
    },
    "metrics": {
        "cross_validation": metricas_cv_final,
        "test": {
            nombre: round(float(valor), 4)
            for nombre, valor in metricas_test.items()
        },
    },
    "versions": versiones_entorno,
}

with ruta_metadata.open(
    mode="w",
    encoding="utf-8",
) as archivo:
    json.dump(
        metadata,
        archivo,
        ensure_ascii=False,
        indent=2,
    )

tamano_modelo_mb = (
    ruta_modelo.stat().st_size / (1024 ** 2)
)

print("Modelo guardado en:")
print(ruta_modelo.resolve())

print("\nMetadatos guardados en:")
print(ruta_metadata.resolve())

print(
    "\nTamaño de model.pkl:",
    round(tamano_modelo_mb, 4),
    "MB",
)
```
